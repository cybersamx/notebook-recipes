# Introduction to Topic Classification

We can apply topic classification to documents or bodies of text to make content generation, organization, and customization more efficient. Topic classification is a broad term, some specific use cases include:

* Product categorization - Analyze and correctly label the products of a company or an eCommerce site into different categories for content organization and personalization.
* News classification - News articles can be identified and categorized into topics for the readers.

## Approaches

In general, topic classification follows this flow where:

A body of text → rules (linguistic, weights, etc) → groups of words (based on the rules)

The groupings become the categories.

In ML-based classification for text, the body of text is broken into a vector of words (feature) with some form of measures (word count, etc) before applying a ML classifier to fit the trained data. 

One downside of this approach is that the standard classification algorithm may be too generic. Although given a large set of quality data may be sufficient for the trained model to be effective. But good training data need to be properly labeled, which can be expensive and slow. For optimal result, perhaps we should include some linguistic and contextual rules into the model. One approach is to automate the labeling using labeling functions that embed the linguistic rules to assign labels to raw data, turning them into labeled data that we can use in training.
