# Using a Simple Sentiment Analysis from Hugging Face

Simple example of using a text classifier from hugging face.



```python
# Install the packages
# Note pytorch must be installed in order for the transformers to work.

!pip install torch transformers > /dev/null 2>&1

import torch
from transformers import pipeline
```

## Simple Sentiment Analysis


```python
# Cuda (only non-mac) and metal (mac silicon) setup.
device = 'cpu:0'
if torch.backends.mps.is_available():
    device = 'mps:0'
if torch.cuda.is_available():
    device = 'cuda:0'

print(f'device={device}')
```

```python
# Load a pipeline using the standard topic classification analysis model. 
topic_classifier = pipeline('zero-shot-classification', device=device)

# Load a pipeline using the standard topic sentiment-analysis model.
sentiment = pipeline('sentiment-analysis', device=device)

# Sample of sentences.
sentences = [
    'This new design of the product is great.',
    'This place serves the best noodles in town.',
    'The service was terrible.',
    'The waiter was rude. I will not come back',
    'I am sad and tired today.',
    'The movie is so so.',
    'Hello',
]

# (Potential) labels for topic classification.
topic_labels = ['food', 'service', 'product', 'movie', 'health']

# Perform topic classification
for sentence in sentences:
    topic_res = topic_classifier(sentence, topic_labels)
    res = sentiment(sentence)

    print(f'Sentence: {sentence}')
    print(f"Topic: {topic_res['labels'][0]} (Score: {topic_res['scores'][0]:.3f})")
    print(f"Sentiment: {res[0]['label']} (Score: {res[0]['score']:.3f})")
    print('-' * 60)
```
